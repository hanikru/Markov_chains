```python
from MarkovC import Markov_chains
from collections import Counter
from pydtmc import MarkovChain
```

1) случайная цепь с 10 состояниями

```python
chain_main = Markov_chains(size = 10)
print(f"вектор стартовых значений\n{chain_main.vec}\n\nпереходная матрица")
for i in range(chain_main.size):
    print(f"{chain_main.mat[i]}")
```

2,3) моделирование значений 

```python
coun = 1000
size_tra = 2

task2 = [chain_main.Mt(size_tra) for _ in range(coun)]
task2_coun = [task2[i][size_tra - 1] for i in range(coun)]
```

```python
mc = MarkovChain(chain_main.mat, states=[(str(i + 1).zfill(4)[:4]) for i in range(chain_main.size)])
start_poz = [i[0] for i in task2]
lib_coun = [mc.simulate(size_tra, initial_state=i-1)[size_tra-1] for i in start_poz]
```

```python
chain_counts = Counter(task2_coun)
chain_counts_lib = Counter(lib_coun)
print("Количество каждой уникальной цепи:")

prob_lib = [count/coun for chain, count in sorted(chain_counts_lib.items())]

for chain, count in sorted(chain_counts.items()):
    prob = 0
    for i in range(chain_main.size):
            prob += chain_main.mat[i][chain - 1]

    prob /= chain_main.size

    print(f"Частота поподания цепи в {chain}: {count/coun:.3f} | вероятность: {prob:.3f} | pydtmc: {prob_lib[chain-1]:.3f}")

```

4, 5) моделирование троекторий сравнение с финальным распределением

```python
coun = 100
size_tra = 100

task4 = [chain_main.Mt(size_tra) for _ in range(coun)]
task4_coun = [task4[i][size_tra - 1] for i in range(coun)]
```

```python
mc = MarkovChain(chain_main.mat, states=[(str(i + 1).zfill(4)[:4]) for i in range(chain_main.size)])
start_poz = [i[0] for i in task4]
lib_coun = [mc.simulate(size_tra, initial_state=i-1)[size_tra-1] for i in start_poz]
```

```python
chain_counts = Counter(task4_coun)
chain_counts_lib = Counter(lib_coun)
print("Количество каждой уникальной цепи:")

prob_lib = [count/coun for chain, count in sorted(chain_counts_lib.items())]

fd = chain_main.Final_distribution()

for chain, count in sorted(chain_counts.items()):
    print(f"Частота поподания цепи в {chain}: {count/coun:.3f} | финальное распределение: {fd[chain-1]:.3f} | pydtmc: {prob_lib[chain-1]:.3f}")
```

6) моделирование случайной траектории

```python
size_tra = 5
task6 = chain_main.Mt(size_tra)

task6_lib = mc.simulate(size_tra, initial_state=task6[0])
task6_lib = [int(i) for i in task6_lib][:-1]

print(f"цепь {task6} | вероятность цепи: {chain_main.Pt(task6):.5f}")
print(f"pydtmc цепь {task6_lib} | вероятность цепи: {chain_main.Pt(task6_lib):.5f}")
```
